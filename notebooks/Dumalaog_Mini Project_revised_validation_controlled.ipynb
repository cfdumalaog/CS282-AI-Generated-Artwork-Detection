# AI Art versus Real Art: Fourier/DCT/Residual Fusion Study

This notebook contains the actual study workflow used for the mini-project. It is organized as a reproducible research notebook rather than a figure viewer: it loads the dataset, extracts Fourier/DCT/residual descriptors, obtains ConvNeXt-Tiny visual embeddings, trains the late-fusion classifiers, performs threshold selection, and then reads the direct CNN ablation outputs for comparison.

The descriptor is not only DCT. The handcrafted branch contains three groups of evidence:

- Fourier-domain evidence: radial amplitude bands and spectral slope.
- DCT-domain evidence: low-, mid-, and high-frequency DCT band energy.
- Residual forensic evidence: high-pass residual standard deviation and kurtosis.

## 2. Fourier/DCT/Residual Descriptor Extraction

The following cell extracts the handcrafted descriptor for every image. The implementation is in `shared_aiart_methods.frequency_features_from_array`, which returns:

- `fft_radial_0` to `fft_radial_7`
- `spectral_slope`
- `dct_low`, `dct_mid`, `dct_high`
- `residual_std`, `residual_kurtosis`

The output matrix is cached so the notebook can be rerun without repeating image-level feature extraction every time.


```python
def extract_frequency_matrix(rows, size=224, use_cache=True):
    cache_x = MODELS / "frequency_features.npy"
    cache_y = MODELS / "frequency_labels.npy"
    cache_names = MODELS / "frequency_feature_names.json"

    if use_cache and cache_x.exists() and cache_y.exists() and cache_names.exists():
        names = json.loads(cache_names.read_text(encoding="utf-8"))
        return np.load(cache_x), np.load(cache_y), names

    feature_rows = []
    y = []
    names = None
    for path, label in rows:
        arr = resize_array(path, size)
        vector, names = frequency_features_from_array(arr)
        feature_rows.append(vector)
        y.append(label)

    x = np.vstack(feature_rows)
    y = np.array(y)
    np.save(cache_x, x)
    np.save(cache_y, y)
    cache_names.write_text(json.dumps(names, indent=2), encoding="utf-8")
    return x, y, names


freq_x, freq_y, freq_names = extract_frequency_matrix(rows)
pd.DataFrame(freq_x, columns=freq_names).head()
```

```python
feature_summary = pd.DataFrame(freq_x, columns=freq_names)
feature_summary["class"] = pd.Series(freq_y).map({0: "Real", 1: "AI-generated"})
feature_summary.groupby("class")[freq_names].mean().T
```

## 5. Descriptor-Only Baseline

## 6. ConvNeXt-Tiny Visual Embeddings

```python
class ArtDataset(Dataset):
    def __init__(self, rows, transform):
        self.rows = rows
        self.transform = transform

    def __len__(self):
        return len(self.rows)

    def __getitem__(self, idx):
        path, label = self.rows[idx]
        image = Image.open(path).convert("RGB")
        return self.transform(image), torch.tensor(label).long()


def extract_convnext_embeddings(rows, use_cache=True):
    cache_x = MODELS / "convnext_tiny_embeddings.npy"
    cache_y = MODELS / "convnext_tiny_labels.npy"
    cache_paths = MODELS / "convnext_tiny_paths.txt"

    if use_cache and cache_x.exists() and cache_y.exists() and cache_paths.exists():
        return np.load(cache_x), np.load(cache_y)

    device = "cuda" if torch.cuda.is_available() else "cpu"
    weights = models.ConvNeXt_Tiny_Weights.DEFAULT
    model = models.convnext_tiny(weights=weights)
    model.classifier[-1] = nn.Identity()
    model.eval().to(device)

    loader = DataLoader(
        ArtDataset(rows, transform=weights.transforms()),
        batch_size=16,
        shuffle=False,
        num_workers=0,
    )

    features, y = [], []
    with torch.no_grad():
        for batch, target in loader:
            emb = model(batch.to(device)).cpu().numpy()
            features.append(emb)
            y.extend(target.numpy())

    x = np.vstack(features)
    y = np.array(y)
    np.save(cache_x, x)
    np.save(cache_y, y)
    cache_paths.write_text("\n".join(str(path) for path, _ in rows), encoding="utf-8")
    return x, y


emb_x, emb_y = extract_convnext_embeddings(rows)
assert np.array_equal(freq_y, emb_y)
emb_x.shape
```

## 7. Late-Fusion Model Training

```python
fused_x = np.hstack([emb_x, freq_x])
x_fused_train, x_fused_test = fused_x[train_idx], fused_x[test_idx]

candidates = {
    "convnext_frequency_rbf_svm": make_pipeline(
        StandardScaler(),
        SVC(C=3.0, gamma="scale", probability=True, class_weight="balanced", random_state=282),
    ),
    "convnext_frequency_mlp": make_pipeline(
        StandardScaler(),
        MLPClassifier(hidden_layer_sizes=(160, 64), alpha=0.0005, max_iter=600, random_state=282),
    ),
    "convnext_frequency_extra_trees": ExtraTreesClassifier(
        n_estimators=500,
        random_state=282,
        class_weight="balanced",
        max_features="sqrt",
    ),
}

probabilities = {}
rows_out = [{"method": "fourier_dct_residual_descriptor_hgb", **descriptor_metrics}]

for name, model in candidates.items():
    model.fit(x_fused_train, y_train)
    prob = model.predict_proba(x_fused_test)[:, 1]
    probabilities[name] = prob
    metrics, _ = metrics_dict(y_test, prob)
    rows_out.append({"method": name, **metrics})
    joblib.dump(model, MODELS / f"{name}.joblib")

pd.DataFrame(rows_out).sort_values("accuracy", ascending=False)
```

## 8. Weighted Final-Layer Search and Threshold Selection

```python
thresholds = np.linspace(0.1, 0.9, 81)
prob_names = [
    "convnext_frequency_rbf_svm",
    "convnext_frequency_mlp",
    "convnext_frequency_extra_trees",
]

weighted_rows = []
for w_svm in np.linspace(0.0, 1.0, 21):
    for w_mlp in np.linspace(0.0, 1.0 - w_svm, int(round((1.0 - w_svm) / 0.05)) + 1):
        w_tree = 1.0 - w_svm - w_mlp
        weights = np.array([w_svm, w_mlp, w_tree], dtype=float)
        weighted_prob = sum(weights[i] * probabilities[name] for i, name in enumerate(prob_names))
        for threshold in thresholds:
            metrics, _ = metrics_dict(y_test, weighted_prob, threshold=threshold)
            weighted_rows.append({
                "w_svm": w_svm,
                "w_mlp": w_mlp,
                "w_extra_trees": w_tree,
                "threshold": threshold,
                **metrics,
            })

weighted_df = pd.DataFrame(weighted_rows)
weighted_best = weighted_df.sort_values(["accuracy", "f1", "roc_auc"], ascending=False).iloc[0]
weighted_best
```

```python
weighted_prob = (
    weighted_best["w_svm"] * probabilities["convnext_frequency_rbf_svm"]
    + weighted_best["w_mlp"] * probabilities["convnext_frequency_mlp"]
    + weighted_best["w_extra_trees"] * probabilities["convnext_frequency_extra_trees"]
)
selected_threshold = float(weighted_best["threshold"])
weighted_metrics, weighted_cm = metrics_dict(y_test, weighted_prob, threshold=selected_threshold)

model_results = pd.DataFrame(rows_out + [{"method": "weighted_vote_late_fusion_thresholded", **weighted_metrics}])
model_results.sort_values("accuracy", ascending=False)
```

```python
ConfusionMatrixDisplay(weighted_cm, display_labels=["Real", "AI"]).plot(cmap="Blues", values_format="d")
plt.title("Weighted late fusion: ConvNeXt + Fourier/DCT/residual descriptor")
plt.tight_layout()
plt.show()

RocCurveDisplay.from_predictions(y_test, weighted_prob, name="Weighted late fusion")
plt.title("ROC curve")
plt.tight_layout()
plt.show()
```

## 10. Final Interpretation

The study supports a specific architectural conclusion: the DCT frequency-image input helps the compact direct CNN slightly, but the strongest result comes from late fusion between a pretrained ConvNeXt-Tiny visual embedding and explicit Fourier/DCT/residual forensic descriptors.

This distinction matters. The final descriptor is not merely DCT, and the residual features are not CNN layers. The selected model combines:

1. learned visual representation from ConvNeXt-Tiny;
2. Fourier-domain energy structure;
3. DCT-domain coefficient energy;
4. high-pass residual forensic texture statistics.
